```python
# ===== Program Utama =====
print("Penyelesaian SPL 3 Variabel (Gauss-Jordan)")
print("Bentuk: a1*x + b1*y + c1*z = d1\n")
```

```python
A = []
b = []
```

```python
def gauss_jordan(A, b):
    n = len(A)
    # Buat matriks augmented [A | b]
    M = [A[i][:] + [b[i]] for i in range(n)]

    for i in range(n):
        # Pivoting: cari baris dengan nilai mutlak terbesar di kolom i
        max_row = max(range(i, n), key=lambda r: abs(M[r][i]))
        if abs(M[max_row][i]) < 1e-12:
            raise ValueError("SPL tidak memiliki solusi tunggal (determinan = 0).")
        M[i], M[max_row] = M[max_row], M[i]

        # Normalisasi baris pivot sehingga elemen pivot = 1
        pivot = M[i][i]
        M[i] = [x / pivot for x in M[i]]

        # Eliminasi semua baris lain (atas dan bawah) pada kolom i
        for j in range(n):
            if j != i:
                faktor = M[j][i]
                M[j] = [M[j][k] - faktor * M[i][k] for k in range(n + 1)]

    return [M[i][n] for i in range(n)]
```

```python
def cetak_matriks(M):
    for baris in M:
        print("  ".join(f"{x:8.3f}" for x in baris))
```

```python
for i in range(3):
    print(f"Persamaan {i+1}:")
    baris = [float(input(f"  Koefisien {v}: ")) for v in ("x", "y", "z")]
    konstanta = float(input("  Konstanta (ruas kanan): "))
    A.append(baris)
    b.append(konstanta)
```

```python
try:
    x, y, z = gauss_jordan(A, b)
    print("\nHasil:")
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
except ValueError as e:
    print("\nError:", e)
```
