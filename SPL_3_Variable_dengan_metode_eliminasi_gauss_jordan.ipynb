# SPL 3 Variabel dengan Eliminasi Gauss-Jordan

**Ide dasar:** ubah matriks augmented `[A | b]` menjadi **matriks identitas** menggunakan Operasi Baris Elementer (OBE). Kolom paling kanan otomatis menjadi jawaban.

$$
\left[\begin{array}{ccc|c}
a_1 & b_1 & c_1 & d_1 \\
a_2 & b_2 & c_2 & d_2 \\
a_3 & b_3 & c_3 & d_3
\end{array}\right]
\xrightarrow{\text{OBE}}
\left[\begin{array}{ccc|c}
1 & 0 & 0 & x \\
0 & 1 & 0 & y \\
0 & 0 & 1 & z
\end{array}\right]
$$

**3 jenis OBE:**
1. Tukar dua baris
2. Kalikan/bagi satu baris dengan angka ≠ 0
3. Kurangi satu baris dengan kelipatan baris lain

**Alur algoritma** (untuk setiap kolom `i` = 0, 1, 2):

`Pivoting` → `Normalisasi` → `Eliminasi`

## Bagian 1 : Fungsi `gauss_jordan(A, b)`

| Tahap | Kode penting | Fungsi |
|-------|--------------|--------|
| Matriks augmented | `M = [A[i][:] + [b[i]] ...]` | Gabungkan koefisien dan konstanta. `[:]` membuat salinan agar `A` asli tidak berubah |
| **Pivoting** | `max(..., key=abs...)` lalu tukar baris | Pilih elemen terbesar sebagai pivot agar tidak membagi dengan 0 dan hasil lebih akurat |
| Cek singular | `< 1e-12` → `raise ValueError` | Jika pivot ≈ 0, SPL tidak punya solusi tunggal |
| **Normalisasi** | `M[i] = [x / pivot ...]` | Bagi seluruh baris pivot (termasuk konstanta) agar pivot = 1 |
| **Eliminasi** | `M[j][k] - faktor * M[i][k]` | Nolkan elemen lain di kolom `i`, **atas dan bawah** |
| Hasil | `M[i][n]` | Kolom terakhir = solusi `[x, y, z]` |

```python
def gauss_jordan(A, b):
    n = len(A)
    # Buat matriks augmented [A | b]
    M = [A[i][:] + [b[i]] for i in range(n)]

    for i in range(n):
        # Pivoting: cari baris dengan nilai mutlak terbesar di kolom i
        max_row = max(range(i, n), key=lambda r: abs(M[r][i]))
        if abs(M[max_row][i]) < 1e-12:
            raise ValueError("SPL tidak memiliki solusi tunggal (determinan = 0).")
        M[i], M[max_row] = M[max_row], M[i]

        # Normalisasi baris pivot sehingga elemen pivot = 1
        pivot = M[i][i]
        M[i] = [x / pivot for x in M[i]]

        # Eliminasi semua baris lain (atas dan bawah) pada kolom i
        for j in range(n):
            if j != i:
                faktor = M[j][i]
                M[j] = [M[j][k] - faktor * M[i][k] for k in range(n + 1)]

    return [M[i][n] for i in range(n)]
```

## Bagian 2 : Input Pengguna

- `A` : list koefisien (3×3)
- `b` : list konstanta ruas kanan
- `input()` membaca teks, `float()` mengubahnya menjadi angka desimal
- `A = []` dan `b = []` ditaruh di sini agar data tidak menumpuk jika sel dijalankan ulang

**Contoh:** `x+y+z=6`, `2x-y+z=3`, `x+2y-z=2` → input `1 1 1 | 6`, `2 -1 1 | 3`, `1 2 -1 | 2`

```python
A = []
b = []

for i in range(3):
    print(f"Persamaan {i+1}:")
    baris = [float(input(f"  Ko1efisien {v}: ")) for v in ("x", "y", "z")]
    konstanta = float(input("  Konstanta (ruas kanan): "))
    A.append(baris)
    b.append(konstanta)
```

## Bagian 3 : Hasil

`try-except` menangkap `ValueError`, sehingga jika SPL tidak punya solusi tunggal, program menampilkan pesan error yang jelas dan tidak crash.

```python
try:
    x, y, z = gauss_jordan(A, b)
    print("\nHasil:")
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
except ValueError as e:
    print("\nError:", e)
```
