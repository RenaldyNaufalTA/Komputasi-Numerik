# Analisis Kode Gauss-Jordan
Gauss-Jordan mengubah sistem persamaan menjadi bentuk yang jawabannya langsung terbaca. Caranya dengan Operasi Baris Elementer (OBE), yaitu tiga hal yang boleh dilakukan pada baris matriks tanpa mengubah solusi:

Menukar dua baris
Mengalikan/membagi satu baris dengan angka tak nol
Mengurangi satu baris dengan kelipatan baris lain

Tujuannya mengubah matriks di sebelah kiri menjadi matriks identitas:

[ 1 0 0 | x ]<br>
[ 0 1 0 | y ]<br>
[ 0 0 1 | z ]<br>

Kolom paling kanan otomatis berisi jawaban.

## Bagian 1 : Deklarasi variabel
Membuat variable array. Variable A untuk menampung nilai koefisien, dan b untuk nilai konstanta.

```python
A = []
b = []
```

### Bagian 2 : Membuat Matriks Augmented
def gauss_jordan(A, b) : Membuat fungsi. A = koefisien, b = konstanta ruas kanan
<br> n = len(A) :	Jumlah persamaan/variabel (di sini 3) <br>
A[i][:] :	Membuat salinan baris ke-i, supaya data asli A tidak ikut berubah <br>
"+ [b[i]]" :	Menempelkan konstanta di ujung baris, sehingga jadi kolom ke-4

```python
def gauss_jordan(A, b):
    n = len(A)
    # Buat matriks augmented [A | b]
    M = [A[i][:] + [b[i]] for i in range(n)]

    for i in range(n):
        # Pivoting: cari baris dengan nilai mutlak terbesar di kolom i
        max_row = max(range(i, n), key=lambda r: abs(M[r][i]))
        if abs(M[max_row][i]) < 1e-12:
            raise ValueError("SPL tidak memiliki solusi tunggal (determinan = 0).")
        M[i], M[max_row] = M[max_row], M[i]

        # Normalisasi baris pivot sehingga elemen pivot = 1
        pivot = M[i][i]
        M[i] = [x / pivot for x in M[i]]

        # Eliminasi semua baris lain (atas dan bawah) pada kolom i
        for j in range(n):
            if j != i:
                faktor = M[j][i]
                M[j] = [M[j][k] - faktor * M[i][k] for k in range(n + 1)]

    return [M[i][n] for i in range(n)]
```

```python
def cetak_matriks(M):
    for baris in M:
        print("  ".join(f"{x:8.3f}" for x in baris))
```

## Bagian 3 : Program Utama (Input dari Pengguna)
input()	Membaca ketikan pengguna (hasilnya teks)
float()	Mengubah teks menjadi bilangan desimal
for v in ("x","y","z")	Meminta koefisien untuk x, y, lalu z
A.append(baris)	Menambahkan satu baris koefisien ke A
b.append(konstanta)	Menyimpan konstanta ruas kanan ke b

```python
for i in range(3):
    print(f"Persamaan {i+1}:")
    baris = [float(input(f"  Koefisien {v}: ")) for v in ("x", "y", "z")]
    konstanta = float(input("  Konstanta (ruas kanan): "))
    A.append(baris)
    b.append(konstanta)
```

## Bagian 4 : Hasil matriks

```python
try:
    x, y, z = gauss_jordan(A, b)
    print("\nHasil:")
    print(f"x = {x:.4f}")
    print(f"y = {y:.4f}")
    print(f"z = {z:.4f}")
except ValueError as e:
    print("\nError:", e)
```
